## streaming and batching

```python
model.invoke("write about c programming language? in 400 words").content
```

```python
 for chunk in model.stream("write about c programming language? in 400 words"):
     print(chunk.content, end="")
```

```python
for chunk in model.stream("why is the sky blue? provide a detailed answer in 500 words reasoning step by step."):
    print(chunk.content, end="",flush=True)    
```

```python
response=model.batch(["What is AI?", "What is machine learning?", "What is deep learning?"])
for r in response:
    print(r.content)
```

```python
response=model.batch(["What is AI?", "What is machine learning?", "What is deep learning?"],config={"stream": True,"max_concurrency": 3
                                                                                                    })
```

```python
for r in response:
    print(r.content)
  
```
